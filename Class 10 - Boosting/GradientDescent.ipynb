```python
# ==========================================
# GRADIENT DESCENT - VISUAL EXPLANATION (FULL VERSION)
# ==========================================

import numpy as np
import matplotlib.pyplot as plt

# 1️⃣ Define the function and its derivative
def f(x):
    return x**2

def df(x):
    return 2*x

# 2️⃣ Parameters
x_start = 8          # Initial position
learning_rate = 0.3  # Step size
n_iterations = 15    # Number of steps

# 3️⃣ Gradient Descent Loop
x = x_start
x_values = [x]
y_values = [f(x)]

for i in range(n_iterations):
    grad = df(x)
    print(f"Step {i:2d}: x = {x:7.4f}, y = {f(x):7.4f}, lr = {learning_rate:2.2f}, , grad = {df(x):2.2f}")
    x = x - learning_rate * grad     # update rule
    x_values.append(x)
    y_values.append(f(x))
    print(f"Conversion {i:2d}: x = {x:7.4f}, y = {f(x):7.4f}")
```

```python
# ==========================================
# 4️⃣ Plot 1: Function and Descent Path
# ==========================================
x_range = np.linspace(-9, 9, 400)
y_range = f(x_range)

plt.figure(figsize=(8,3))
# --- Left: Function and Descent Path
plt.subplot(1,2,1)
plt.plot(x_range, y_range, 'b-', label='f(x) = x²')

plt.title('Gradient Descent Space')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(alpha=0.3)
```

```python
# ==========================================
# 4️⃣ Plot 1.1: Function and Descent Path
# ==========================================
x_range = np.linspace(-9, 9, 400)
y_range = f(x_range)

plt.figure(figsize=(8,3))
# --- Left: Function and Descent Path
plt.subplot(1,2,1)
plt.plot(x_range, y_range, 'b-', label='f(x) = x²')
plt.scatter(x_values, y_values, c='red', s=60, zorder=5)
plt.plot(x_values, y_values, 'r--', label='Gradient Descent path')

# Annotate steps
for i, (x_i, y_i) in enumerate(zip(x_values, y_values)):
    plt.text(x_i, y_i+2, f'{i}', fontsize=8, ha='center')

plt.title('Gradient Descent – Step by Step')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(alpha=0.3)
```

```python
# ==========================================
# 5️⃣ Plot 2: Loss per Iteration
# ==========================================
plt.subplot(1,2,2)
plt.plot(range(len(y_values)), y_values, 'g-o')
plt.title('Loss Decrease per Iteration')
plt.xlabel('Iteration')
plt.ylabel('f(x)')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ==========================================
# 6️⃣ Print Results
# ==========================================
print("Iteration Results:")
for i, (x_i, y_i) in enumerate(zip(x_values, y_values)):
    print(f"Step {i:2d}: x = {x_i:7.4f}, f(x) = {y_i:7.4f}")
```
